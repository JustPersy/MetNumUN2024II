## The variables: Include new variables which information that you believe could improve the price prediction, in a comment explain your choice.

Decidí incluir las **moving averages (medias móviles)** como una nueva variable en el conjunto de características debido a su capacidad para identificar tendencias en el mercado y suavizar las fluctuaciones a corto plazo, lo que es especialmente útil en la predicción de precios.

Las medias móviles proporcionan una perspectiva clara de la dirección del precio (al alza, a la baja o lateral) al reducir el impacto del ruido del mercado. En particular, utilicé:
- **`sma_slow`**: Una media móvil simple a 20 períodos que captura tendencias de mediano plazo.
- **`sma_fast`**: Una media móvil simple a 5 períodos que detecta cambios más rápidos en los precios.

La relación entre estas dos medias (lenta y rápida) es valiosa para identificar cruces de medias móviles, que a menudo se consideran señales de compra (cuando la rápida cruza por encima de la lenta) o de venta (cuando la rápida cruza por debajo de la lenta). Este comportamiento se codifica en la variable **`weights`**, asignando un valor de 1 o -1 según la dirección de la tendencia detectada.

La inclusión de estas medias móviles mejora el modelo al:
1. **Capturar tendencias fundamentales**: Esto es esencial para predecir precios en mercados con movimientos direccionales.
2. **Proporcionar señales robustas**: Minimiza el impacto del ruido en los datos y ayuda a identificar momentos clave de entrada o salida del mercado.
3. **Complementar otras características**: Como el MACD y el RSI, las medias móviles agregan una capa adicional de información basada en la relación entre precios recientes y pasados.



## Objetivo a predecir

Decidí implementar esta estrategia porque simplifica la toma de decisiones en el trading al clasificar las posibles acciones en tres categorías claras: comprar, vender o mantener la posición. Este enfoque utiliza el cambio porcentual entre el precio actual y el futuro como métrica clave, lo que permite identificar oportunidades de compra cuando se espera un aumento en el precio y mitigar pérdidas al recomendar ventas solo ante caídas significativas, definidas por un umbral configurable. Además, al incluir la opción de mantener la posición, se evita realizar operaciones innecesarias que podrían generar costos adicionales. La estrategia es flexible y fácilmente ajustable, lo que me permite adaptarla a distintos contextos del mercado y niveles de volatilidad, mientras mantiene un enfoque en la optimización y protección del capital.
